```python
import pandas as pd
import numpy as np

# ============================================================
# 1. SETTINGS
# ============================================================

np.random.seed(42)

total_records = 20000
original_records = 19000
duplicate_records = 1000


# ============================================================
# 2. MASTER DATA
# ============================================================

first_names = [
    'Rahul', 'Amit', 'Rohit', 'Vikas', 'Suresh',
    'Ankit', 'Arjun', 'Karan', 'Vivek', 'Raj',
    'Priya', 'Neha', 'Pooja', 'Anjali', 'Sneha',
    'Kavita', 'Riya', 'Shreya', 'Swati', 'Nisha',
    'Akash', 'Manish', 'Abhishek', 'Deepak', 'Sanjay'
]

last_names = [
    'Sharma', 'Kumar', 'Patel', 'Singh', 'Das',
    'Panda', 'Reddy', 'Rao', 'Mishra', 'Verma',
    'Gupta', 'Mehta', 'Jain', 'Nayak', 'Sahu',
    'Behera', 'Mohanty', 'Dutta', 'Yadav', 'Pradhan'
]

products = [
    'Laptop',
    'Smartphone',
    'Tablet',
    'Headphones',
    'Smart Watch',
    'Keyboard',
    'Mouse',
    'Monitor',
    'Printer',
    'Camera',
    'Power Bank',
    'Earbuds',
    'Bluetooth Speaker',
    'Gaming Chair',
    'Backpack',
    'USB Cable',
    'External Hard Drive',
    'SSD',
    'Webcam',
    'Router'
]

categories = [
    'Electronics',
    'Computer Accessories',
    'Furniture',
    'Bags',
    'Networking'
]

cities = [
    'Hyderabad',
    'Bangalore',
    'Mumbai',
    'Delhi',
    'Pune',
    'Chennai',
    'Kolkata',
    'Bhubaneswar',
    'Cuttack',
    'Visakhapatnam',
    'Ahmedabad',
    'Jaipur',
    'Lucknow',
    'Bhopal',
    'Patna'
]

states = [
    'Telangana',
    'Karnataka',
    'Maharashtra',
    'Delhi',
    'Maharashtra',
    'Tamil Nadu',
    'West Bengal',
    'Odisha',
    'Odisha',
    'Andhra Pradesh',
    'Gujarat',
    'Rajasthan',
    'Uttar Pradesh',
    'Madhya Pradesh',
    'Bihar'
]

payment_methods = [
    'UPI',
    'Credit Card',
    'Debit Card',
    'Cash on Delivery',
    'Net Banking',
    'Wallet'
]

order_status = [
    'Delivered',
    'Delivered',
    'Delivered',
    'Delivered',
    'Cancelled',
    'Returned',
    'Pending'
]


# ============================================================
# 3. GENERATE 19,000 ORIGINAL RECORDS
# ============================================================

data = []

for i in range(original_records):

    first_name = np.random.choice(first_names)
    last_name = np.random.choice(last_names)

    customer_name = first_name + ' ' + last_name

    order_id = f'ORD{100001 + i}'

    customer_id = f'CUST{10001 + i}'

    gender = np.random.choice([
        'Male',
        'Female'
    ])

    age = np.random.randint(
        18,
        66
    )

    product = np.random.choice(
        products
    )

    category = np.random.choice(
        categories
    )

    city_index = np.random.randint(
        0,
        len(cities)
    )

    city = cities[city_index]

    state = states[city_index]

    quantity = np.random.randint(
        1,
        6
    )

    unit_price = np.random.randint(
        500,
        100000
    )

    discount = np.random.choice([
        0,
        5,
        10,
        15,
        20,
        25,
        30
    ])

    sales = (
        quantity *
        unit_price *
        (1 - discount / 100)
    )

    sales = round(
        sales,
        2
    )

    profit = round(
        sales * np.random.uniform(
            0.05,
            0.25
        ),
        2
    )

    order_date = pd.Timestamp(
        np.random.choice(
            pd.date_range(
                '2022-01-01',
                '2025-12-31'
            )
        )
    )

    payment = np.random.choice(
        payment_methods
    )

    status = np.random.choice(
        order_status
    )

    data.append([
        order_id,
        order_date,
        customer_id,
        customer_name,
        gender,
        age,
        product,
        category,
        quantity,
        unit_price,
        discount,
        sales,
        profit,
        payment,
        city,
        state,
        status
    ])


# ============================================================
# 4. CREATE DATAFRAME
# ============================================================

columns = [
    'Order_ID',
    'Order_Date',
    'Customer_ID',
    'Customer_Name',
    'Gender',
    'Age',
    'Product',
    'Category',
    'Quantity',
    'Unit_Price',
    'Discount_Percentage',
    'Sales',
    'Profit',
    'Payment_Method',
    'City',
    'State',
    'Order_Status'
]

df = pd.DataFrame(
    data,
    columns=columns
)


# ============================================================
# 5. ADD MISSING VALUES INTENTIONALLY
# ============================================================

# Customer Name - 150 missing
index = np.random.choice(
    df.index,
    150,
    replace=False
)

df.loc[index, 'Customer_Name'] = np.nan


# Gender - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Gender'] = np.nan


# Age - 120 missing
index = np.random.choice(
    df.index,
    120,
    replace=False
)

df.loc[index, 'Age'] = np.nan


# Product - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Product'] = np.nan


# Category - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Category'] = np.nan


# Quantity - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Quantity'] = np.nan


# Unit Price - 120 missing
index = np.random.choice(
    df.index,
    120,
    replace=False
)

df.loc[index, 'Unit_Price'] = np.nan


# Discount - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Discount_Percentage'] = np.nan


# Sales - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Sales'] = np.nan


# Profit - 120 missing
index = np.random.choice(
    df.index,
    120,
    replace=False
)

df.loc[index, 'Profit'] = np.nan


# Payment Method - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Payment_Method'] = np.nan


# City - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'City'] = np.nan


# State - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'State'] = np.nan


# Order Status - 100 missing
index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Order_Status'] = np.nan


# ============================================================
# 6. ADD INCONSISTENT DATA INTENTIONALLY
# ============================================================

# Category inconsistencies

index = np.random.choice(
    df.dropna(subset=['Category']).index,
    300,
    replace=False
)

category_changes = [
    'electronics',
    'ELECTRONICS',
    ' Electronics ',
    'computer accessories',
    'COMPUTER ACCESSORIES',
    ' Computer Accessories ',
    'furniture',
    'FURNITURE',
    ' Furniture ',
    'bags',
    'BAGS',
    ' Bags '
]

for i in index:
    df.loc[i, 'Category'] = np.random.choice(
        category_changes
    )


# City inconsistencies

index = np.random.choice(
    df.dropna(subset=['City']).index,
    300,
    replace=False
)

city_changes = [
    'hyderabad',
    'HYDERABAD',
    ' Hyderabad ',
    'bangalore',
    'BANGALORE',
    ' Bangalore ',
    'mumbai',
    'MUMBAI',
    ' Mumbai ',
    'delhi',
    'DELHI',
    ' Delhi ',
    'pune',
    'PUNE',
    ' Pune '
]

for i in index:
    df.loc[i, 'City'] = np.random.choice(
        city_changes
    )


# ============================================================
# 7. ADD INVALID VALUES INTENTIONALLY
# ============================================================

# Invalid Quantity

index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Quantity'] = np.random.choice([
    -5,
    -1,
    0,
    100,
    500
], 100)


# Invalid Discount

index = np.random.choice(
    df.index,
    100,
    replace=False
)

df.loc[index, 'Discount_Percentage'] = np.random.choice([
    -10,
    -5,
    105,
    150
], 100)


# Invalid Sales / Outliers

index = np.random.choice(
    df.index,
    50,
    replace=False
)

df.loc[index, 'Sales'] = np.random.choice([
    -500,
    -1000,
    500000,
    1000000
], 50)


# ============================================================
# 8. ADD 1,000 DUPLICATE RECORDS INTENTIONALLY
# ============================================================

duplicate_data = df.sample(
    n=duplicate_records,
    random_state=42
)

df = pd.concat(
    [
        df,
        duplicate_data
    ],
    ignore_index=True
)


# ============================================================
# 9. SHUFFLE THE DATA
# ============================================================

df = df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)


# ============================================================
# 10. CHECK EXACT NUMBER OF RECORDS
# ============================================================

print("Number of records:", len(df))

print("Number of columns:", len(df.columns))


# ============================================================
# 11. DISPLAY DATAFRAME
# ============================================================

df.head()
```
